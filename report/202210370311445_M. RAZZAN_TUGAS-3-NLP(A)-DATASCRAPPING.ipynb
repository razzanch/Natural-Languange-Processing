## Bagian 2: Pra-Pemrosesan Data

Setelah mengumpulkan data, langkah selanjutnya adalah membersihkan dan melakukan pra-pemrosesan. Hal ini meliputi:
1. Mengubah teks menjadi huruf kecil
2. Menghapus tanda baca
3. Menghapus stopwords
4. Melakukan tokenisasi teks

Saya akan menggunakan NLTK (Natural Language Toolkit) untuk tugas-tugas pra-pemrosesan ini.

```python
import csv
import re
import nltk
from nltk.corpus import stopwords
import string

# Unduh sumber daya NLTK jika belum ada
nltk.download('stopwords')
nltk.download('punkt')
```

```python
def clean_csv_and_preprocess(input_file, output_file):
    # Memuat stopwords
    stop_words = set(stopwords.words('english'))
    
    def preprocess_text(text):
        # Mengubah teks menjadi huruf kecil
        text = text.lower()
        
        # Menghapus tanda baca
        text = text.translate(str.maketrans('', '', string.punctuation))
        
        # Tokenisasi teks
        tokens = nltk.word_tokenize(text)
        
        # Menghapus stopwords dan token non-alfabetik
        tokens = [word for word in tokens if word not in stop_words and word.isalpha()]
        
        # Menggabungkan token kembali menjadi string
        return " ".join(tokens)
    
    fixed_rows = []
    
    # Meningkatkan batas ukuran field untuk field teks besar
    csv.field_size_limit(1000000)
    
    with open(input_file, 'r', encoding='utf-8') as file:
        reader = csv.reader(file)
        header = next(reader)  # Lewati header
        
        current_url = None
        current_text = ""
        
        for row in reader:
            if len(row) >= 2:  # Pastikan baris memiliki minimal dua kolom
                url, text = row[0], row[1]
                
                # Periksa apakah kolom pertama adalah URL yang valid
                if re.match(r'^https?://', url):
                    # Jika kita sudah memiliki URL valid dan teks, simpan
                    if current_url and current_text.strip():
                        fixed_rows.append([current_url, current_text.strip()])
                    
                    # Mulai entri baru
                    current_url = url
                    current_text = text
                else:
                    # Tambahkan ke teks saat ini jika URL tidak valid
                    current_text += " " + url + " " + text
        
        # Tambahkan entri terakhir
        if current_url and current_text.strip():
            fixed_rows.append([current_url, current_text.strip()])
    
    # Pra-pemrosesan sampel teks
    preprocessed_rows = []
    for url, text in fixed_rows:
        cleaned_text = preprocess_text(text)
        preprocessed_rows.append([url, cleaned_text])
    
    # Simpan data yang dibersihkan dan dipra-pemrosesan ke file CSV baru
    with open(output_file, 'w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow(["URL", "Teks yang Dibersihkan"])
        writer.writerows(preprocessed_rows)
    
    print(f"Data telah dibersihkan dan dipra-pemrosesan. Tersimpan ke {output_file}")
```

```python
# Jalankan fungsi pra-pemrosesan
input_csv = "merged_TextSample.csv"  # Menggunakan dataset gabungan
output_csv = "cleaned_webdev_text_samples.csv"
clean_csv_and_preprocess(input_csv, output_csv)
```

## Analisis Data dan Ringkasan

Mari menganalisis data yang telah dikumpulkan dan dipra-pemrosesan untuk mendapatkan wawasan.

```python
import pandas as pd
from urllib.parse import urlparse
from collections import Counter

# Memuat data yang telah dibersihkan
df = pd.read_csv("cleaned_webdev_text_samples.csv")

# Menghitung rata-rata panjang teks (dalam kata)
df['word_count'] = df['Cleaned Text'].apply(lambda x: len(str(x).split()))
avg_length = df['word_count'].mean()

# Ekstrak domain dari URL
df['domain'] = df['URL'].apply(lambda x: urlparse(x).netloc)
domains = Counter(df['domain'])
top_domains = domains.most_common(5)

# Cetak statistik ringkasan
print("Analisis Data Hasil Scraping (Data telah dibersihkan dan dipra-pemrosesan):")
print("------------------------------------------------------------------------------")
print(f"Total sampel: {len(df)}")
print(f"Rata-rata panjang teks (dalam kata): {int(avg_length)}")
print(f"Total situs web unik yang diambil: {len(domains)}")
print("\nDomain yang paling umum dalam dataset:")
for i, (domain, count) in enumerate(top_domains, 1):
    print(f"{i}. {domain}: {count} artikel")
```
